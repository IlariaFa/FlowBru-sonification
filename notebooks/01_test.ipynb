Objectif : faire varier la hauteur du son selon ces valeurs.

```python
donnees = [0.2, 0.5, 0.9, 0.4, 0.7]
```

Importer les biblpiotheques

```python
import numpy as np
import matplotlib.pyplot as plt

from strauss.sonification import Sonification
from strauss.sources import Objects
from strauss.score import Score
from strauss.generator import Synthesizer
```

Créer et visualiser les données 

```python
donnees = np.array([0.2, 0.5, 0.9, 0.4, 0.7])

plt.plot(donnees, marker='o')
plt.xlabel("Point de données")
plt.ylabel("Valeur")
plt.title("Mes premières données")
plt.show()
```

Préparer les données pour STRAUSS:
Ici :
- len(donnees) compte les cinq valeurs.
- x indique leur progression dans le temps.
- pitch_shift reçoit les valeurs qui contrôleront la hauteur.
- azimuth et polar définissent une position spatiale fixe.

```python
x = np.linspace(0, 1, len(donnees))

data = {
    'pitch': 1.,
    'time_evo': x,
    'azimuth': 0.5,
    'polar': 0.5,
    'pitch_shift': donnees
}
```

Generer le son

```python
length = 5.
system = "stereo"

notes = [["A2"]]
score = Score(notes, length)

generator = Synthesizer()
generator.load_preset("pitch_mapper")

sources = Objects(data.keys())
sources.fromdict(data)
sources.apply_mapping_functions()

soni = Sonification(score, sources, generator, system)
soni.render()
soni.notebook_display()
```
